```python
#Dictionary Properties
 #Key- Value Structure
#Data is stored ad: 
#Key : Value

#Example ----

student = {
    "Name": "Ravi",
    "Age": 24
}
print(student)
```

```python
#Mutable - Dictionaries are mutable
#We can :
'''
 # 1. Add key-value pairs
   2. Update existing value
   3. Remove entries
'''
#Example ---

student = {
    "Name" : "Ravi",
    "Age" : 24
}
student["Age"] = 25  # update the age of student
print(student)
```

```python
#Duplicate Keys - a dictonary cannot meaningfully contain duplicate key 
# if the same key appears again the latest value overwrite the previous value

student = {
    "Name" : "Ravi",
    "Name" : "Amit" #previous value is overwrite because of duplicate
}
print(student)
```

```python
#Accessing Dictionary values - Use the key

info = {
    "name" : "Apna College",
    "Age" : 35
}
print(info["name"])
print(info["Age"])
# dictionary access is dict[key] not by dict[index]
```

```python
#KeyError - if you access a key that dosent exist
student = {
    "Name" : "Rahul"
}
print(student["age"]) #keyError this is an important interview and debugging concept
```

```python
#Updating and Existing Key - assign a new value to an existing key
info ={
    "Name" : "Apna College",
    "age" : 35
}
info["Name"] = "Apni Duniya" # the Existing Value Replaced
print(info)
```

```python
#Adding a New Key -  if the key doesnt exit assignment create a new key- value pair

info ={
    "Name" : "Ravi"
}
info["Surname" ]= "Verma"
print(info)
#Golden Rule
'''
dict[key] = value
 >Existing key -> update
 >New key -> add
 '''
```

```python
#Create an empty dictionary using Empty dict
null_dict = {}  
print(null_dict)
print(type(null_dict))
null_dict["Name"] = "Ravi"
null_dict["Age" ] = 25
print(null_dict)

# means an empty dictonary not an empty set
```

```python
#Nested Dictonary can contain another dictoonary as a value
#This is called a nested dictionary.
 
student = {
    "Name" : "Rahul Kumar",
    "Subject" :{
        "Physic" : 55,
        "Chemistry" : 67,
        "Math" : 95
    }
}
print(student)
```

```python
#Accessing Nested ditionary Data -  
print(student["Subject"]) #Access the complete inner dictionary
print(student["Subject"]["Chemistry"]) #Access a specfic subject
```

```python
#Dictionary Method - Key() 
#Return a view containing all dictionary keys

student = {
    "Name" : "Rahul",
    "Subject" : {
        "Physics" : 80,
        "Chemistry" : 90
    }
}
print(student.keys())
print(len(student.keys()))
```
