```python
#Dictionary Properties
 #Key- Value Structure
#Data is stored ad: 
#Key : Value

#Example ----

student = {
    "Name": "Ravi",
    "Age": 24
}
print(student)
```

```python
#Mutable - Dictionaries are mutable
#We can :
'''
 # 1. Add key-value pairs
   2. Update existing value
   3. Remove entries
'''
#Example ---

student = {
    "Name" : "Ravi",
    "Age" : 24
}
student["Age"] = 25  # update the age of student
print(student)
```

```python
#Duplicate Keys - a dictonary cannot meaningfully contain duplicate key 
# if the same key appears again the latest value overwrite the previous value

student = {
    "Name" : "Ravi",
    "Name" : "Amit" #previous value is overwrite because of duplicate
}
print(student)
```

```python
#Accessing Dictionary values - Use the key

info = {
    "name" : "Apna College",
    "Age" : 35
}
print(info["name"])
print(info["Age"])
# dictionary access is dict[key] not by dict[index]
```

```python
#KeyError - if you access a key that dosent exist
student = {
    "Name" : "Rahul"
}
print(student["age"]) #keyError this is an important interview and debugging concept
```

```python
#Updating and Existing Key - assign a new value to an existing key
info ={
    "Name" : "Apna College",
    "age" : 35
}
info["Name"] = "Apni Duniya" # the Existing Value Replaced
print(info)
```

```python
#Adding a New Key -  if the key doesnt exit assignment create a new key- value pair

info ={
    "Name" : "Ravi"
}
info["Surname" ]= "Verma"
print(info)
#Golden Rule
'''
dict[key] = value
 >Existing key -> update
 >New key -> add
 '''
```

```python
#Nested Dictonary can contain another dictoonary as a value
#This is called a nested dictionary.
 
student = {
    "Name" : "Rahul Kumar",
    "Subject" :{
        "Physic" : 55,
        "Chemistry" : 67,
        "Math" : 95
    }
}
print(student)
```

```python
#SET -- set  a set is an unordered  collection of unique element It is similar to the mathematical concept of a set
collection = {
    1,1,2,3,4,5,"hello","World"
}
print(type(collection))
print(collection)
```

```python
#Set Properties Sets automatically remove duplicate
s={
    9,1,2,2,3,5,"Hello","Hello"
}
print(s)
#the repeated value are stored only once
```

```python
#Unorderd Set do not provide index based access 
#You should not depend on the displayed order of set element 

s = {10,20,30} 
s[0] # set are designed around membership and uniqueness not positional indexing
```

```python
#Set lenght Use len():  it will ignor the duplicate
s = { 1,2,3,4,5,5}
print(len(s))
```

```python
#Empty set
empty = set()
print(type(empty))
```

```python
#Set Mutability vs Element Immutability
s = {1,2} # set themselves are mutable collection you can add remove element
s.add(3)
s.add(4)
print(s)
s.remove(4)
print(s)
#Element stored inside a set must be hashable
#it allowed int, float, string, boolean,  but not allowed list[1,2] dictionary{"a":1}
```

```python
#Set Method - add(element)
s= {1,2}
s.add(3)
print(s)
s.add(2) #duplicate add and set automatically maintain uniquness
print(s)
```

```python
#Set Method -- remove(element) remvove an element form a set 
s = {1,2,3,4}
s.remove(1)
print(s)
s.remove(6)
print(s)
```

```python
#Diffrent Data type in sets
# A set contain multiple hashable types
s = set()
s.add("Hello")
s.add((1,2,3))
s.add(100)
s.add(3.34)
print(s)
s.add[2,3] # is invlaid because a list is unhashable
print(s)
```
