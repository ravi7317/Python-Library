# Q.7 Show all the record where category is movie and type is comedies is united kingdom

```python
import pandas as pd

data = pd.read_csv(r"D:\Pands Python\Project\Netflix-Data\8. Netflix Dataset.csv")

result = data[
    (data["Category"].str.lower() == "movie")
    & data["Type"].str.contains("comedies", case=False, na=False)
    & data["Country"].str.contains("United Kingdom", case=False, na=False)
]

result
```

```python
genre_counts = (
    result["Type"]
    .str.split(", ")
    .explode()
    .value_counts()
)

genre_counts.plot(kind="bar", title="Genre tags in UK comedy movies", ylabel="Number of titles")
```

```python
# Number of movies by rating
result["Rating"].value_counts().plot(
    kind="bar",
    title="UK Comedy Movies by Rating",
    xlabel="Rating",
    ylabel="Number of movies"
)

# Movies by release year
release_years = pd.to_datetime(
    result["Release_Date"], format="mixed", errors="coerce"
).dt.year

release_years.value_counts().sort_index().plot(
    kind="bar",
    title="UK Comedy Movies by Release Year",
    xlabel="Year",
    ylabel="Number of movies"
)

# Directors with the most movies
result["Director"].value_counts().head(10)

# Movie duration statistics (minutes)
durations = pd.to_numeric(
    result["Duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
)
durations.describe()

# List titles alphabetically
result[["Title", "Director", "Release_Date", "Rating"]].sort_values("Title")
```

```python
country_counts = (
    data["Country"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

country_counts.plot(
    kind="bar",
    title="Top 10 Countries by Number of Netflix Titles",
    xlabel="Country",
    ylabel="Number of titles",
    rot=45
)
```

```python
# Number of Netflix titles by rating
data["Rating"].value_counts().plot(
    kind="bar",
    title="Netflix Titles by Rating",
    xlabel="Rating",
    ylabel="Number of titles",
    rot=45
)
```
